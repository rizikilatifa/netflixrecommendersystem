## Popularity Based Reccomensation System


```python
import pandas as pd
import numpy as np
```

```python
movie_names = pd.read_csv("movies.csv")
movie_names.head()
```

```python
movie_names.shape
```

```python
movie_ratings = pd.read_csv("ratings.csv")
movie_ratings.head()
movie_ratings.shape
```

```python
movie_data = pd.merge(movie_ratings, movie_names, on='movieId')
movie_data.head()
```

```python
movie_data.shape
```

## Criteria for Popularity Based Reccommendation Systems
1. Movies with the highest rating
2. Number of Views

### Average ratings of movies

```python
movie_data.groupby('title')['rating'].mean().head()
```

```python
# sorting them with movies wwith highest ratings
movie_data.groupby('title')['rating'].mean().sort_values(ascending=False).head(30)
```

### Rating Count

```python
movie_data.groupby('title')['rating'].count().sort_values(ascending=False).head()
```

```python
movie_data[['title', 'rating']]
```

```python
# calculate the average ratings
ratings_mean_count = pd.DataFrame(movie_data.groupby('title')['rating'].mean())
```

```python
ratings_mean_count
```

### calculate the count of the rating and add to the average ratings above

```python
ratings_mean_count['rating_count'] = movie_data.groupby('title')['rating'].count()
```

```python
ratings_mean_count
```

```python
ratings_mean_count['rating'] = round(ratings_mean_count['rating'],1)
```

```python
ratings_mean_count.tail()
```

```python
ratings_mean_count = ratings_mean_count[(ratings_mean_count['rating'] > 3) & (ratings_mean_count['rating_count'] > 100) ]
```

```python
ratings_mean_count.head()
```

```python
ratings_mean_count = ratings_mean_count.sort_values(by = 'rating', ascending=False)
```
